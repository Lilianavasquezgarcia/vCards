Automated generation of vCards for multiple contact lists

```python
import csv

input_csv = 'Contacto_EMPRENDEDORES.csv'   # upload data Name, phone and other infromation if necesary
output_vcf = 'contactos.vcf'

with open(input_csv, mode='r', encoding='latin-1') as file:
    reader = csv.DictReader(file, delimiter=';')

    with open(output_vcf, mode='w', encoding='utf-8') as vcf:
        for row in reader:
            name = row.get('Name', '').strip()
            prefijo = row.get('Prefijo', '').strip()
            phone = row.get('Phone', '').strip()
            org = row.get('Organization Name', '').strip()
            title = row.get('Organization Title', '').strip()

            # Unir prefijo y teléfono de manera limpia
            full_phone = f"+{prefijo}{phone}" if prefijo and not phone.startswith('+') else phone

            vcf.write("BEGIN:VCARD\n")
            vcf.write("VERSION:3.0\n")
            if name:
                vcf.write(f"FN:{name}\n")
                vcf.write(f"N:;{name};;;\n")
            if full_phone:
                vcf.write(f"TEL;TYPE=CELL:{full_phone}\n")
            if org:
                vcf.write(f"ORG:{org}\n")
            if title:
                vcf.write(f"TITLE:{title}\n")
            vcf.write("END:VCARD\n")

print("vCard generated successfully!")
```
